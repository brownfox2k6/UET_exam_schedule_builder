### General configurations & Hyperparamters

```python
# Ant Colony Optimization
MAX_ITERS   = 200     # Maximum number of iterations
NUM_ANTS    = 40       # Number of ants used in each iteration
ALPHA       = 1.0      # Pheromone importance in selection probability (τ^α)
BETA        = 3.0      # Heuristic importance in selection probability (η^β)
RHO         = 0.08     # Pheromone evaporation rate for global update
XI          = 0.0      # Local pheromone decay rate after each step
TAU_MIN     = 0.003    # Minimum pheromone trail limit
TAU_MAX     = 1.0      # Maximum pheromone trail limit
TAU_0       = 1.0      # Initial pheromone level (uniformly distributed)
GREEDY_PROB = 0.0      # Probability of an ant greedily choosing the best slot instead of probabilistic selection

# Local Search
TOP_ANTS    = 0.1      # Ratio of top-performing ants selected for Local Search refinement
MOVE_PROB   = 0.7      # Probability of applying 1-move operator instead of 2-swap
MAX_MOVES   = 200      # Maximum number of actual successful moves (downhill steps)
MAX_FAILS   = 2000     # Stop after this many consecutive failed attempts

# Stagnation & Convergence
STAGNATION_LIMIT = 5000  # Max consecutive iterations without improvement before applying reset/perturbation
```

### Tạo danh sách ngày thi và ca thi

```python
SLOTS_IN_DAY = [time(8), time(10, 30), time(13), time(15, 30), time(18)]  # Các ca thi trong ngày
SCHEDULE_ON_SATURDAY = True    # Có tổ chức thi vào Thứ bảy không?
SCHEDULE_ON_SUNDAY   = False   # Có tổ thức thi vào Chủ nhật không?
DATE_START = date(2025, 8, 18)  # inclusive
DATE_END   = date(2025, 8, 29)  # inclusive
TERM_ID    = "243"   # Mã học kỳ

SLOTS: list[datetime] = []
_date = DATE_START
while _date <= DATE_END:
  is_off = (_date.weekday() == 5 and not SCHEDULE_ON_SATURDAY) or \
           (_date.weekday() == 6 and not SCHEDULE_ON_SUNDAY)
  if not is_off:
    SLOTS.extend(datetime.combine(_date, slot) for slot in SLOTS_IN_DAY)
  _date += timedelta(days=1)

_epoch = datetime.combine(DATE_START, time())
ABSOLUTE_DAY_SLOTS = np.array([
  (slot - _epoch).total_seconds() / 86400
  for slot in SLOTS
], dtype=np.float64)

NUM_SLOTS = len(SLOTS)
print(f"Số ca thi một ngày: {len(SLOTS_IN_DAY)}")
print(f"Số ngày thi: {NUM_SLOTS // len(SLOTS_IN_DAY)}")
print(f"Số ca thi: {NUM_SLOTS}")
```

### Xây dựng ma trận xung đột `conflict`
- `conflict[c1, c2]` cho biết có bao nhiêu người cùng học 2 môn `c1` và `c2`
- `total_conflict[c]` cho biết tổng quy mô xung đột của môn `c` so với toàn bộ phần còn lại.

```python
_courses_by_student = defaultdict(list)
for e in ENROLMENTS:
  _courses_by_student[e.student_id].append(COURSES.index(e.course_id))

CONFLICT = np.zeros((NUM_COURSES, NUM_COURSES), dtype=np.int32)
for student_courses in _courses_by_student.values():
  for c1, course2 in itertools.combinations(student_courses, 2):
    CONFLICT[c1, course2] += 1
    CONFLICT[course2, c1] += 1

TOTAL_CONFLICT = np.sum(CONFLICT, axis=1).astype(np.int32)

_nonzero = np.count_nonzero(CONFLICT)
print(f"Kích thước ma trận conflict: {CONFLICT.size} ({NUM_COURSES} x {NUM_COURSES})")
print(f"Số phần tử khác 0: {_nonzero} ({100 * _nonzero / CONFLICT.size:.2f}%)")
```

### Heuristic: Tạo thứ tự xếp lịch thi
Tìm môn học tiếp theo để xếp lịch dựa trên Heuristic: Saturation Degree (Độ bão hoà).  
Ưu tiên môn có số lượng ca thi **khả thi ít nhất** (nói cách khác, **bị cấm nhiều nhất**).  
Nếu có nhiều môn cùng thoả mãn, chọn môn có tổng quy mô xung đột (với toàn bộ các môn khác) lớn nhất.

```python
@njit
def _get_next_course(
  schedule: NDArray[np.int32],
  forbidden_count: NDArray[np.int32],
  num_courses: int = NUM_COURSES,
  total_conflict: NDArray[np.int32] = TOTAL_CONFLICT
) -> int:

  best_course = max_degree = max_conflict = -1

  for c in range(num_courses):
    # Bỏ qua nếu đã được xếp lịch
    if schedule[c] != -1:
      continue

    # Cập nhật môn tốt nhất (môn có nhiều slot bị cấm nhất)
    degree = forbidden_count[c]
    conflict = total_conflict[c]
    if ((degree > max_degree)
        or (degree == max_degree and conflict > max_conflict)):
      max_degree = degree
      max_conflict = conflict
      best_course = c
      
  return best_course
```

```python
@njit
def _update_forbidden_mask(
  newly_assigned_course: int,
  chosen_slot: int,
  schedule: NDArray[np.int32],
  forbidden_mask: NDArray[np.bool_],
  forbidden_count: NDArray[np.int32],
  num_courses: int = NUM_COURSES,
  conflict: NDArray[np.int32] = CONFLICT
) -> None:

  for c in range(num_courses):
    # Nếu môn đang xét có xung đột và chưa được xếp lịch
    if schedule[c] == -1 and conflict[newly_assigned_course, c] > 0:

      # Nếu ca thi này chưa bị cấm từ trước
      if not forbidden_mask[c, chosen_slot]:
        forbidden_mask[c, chosen_slot] = True
        forbidden_count[c] += 1
```

### Xây đường đi (lịch thi) cho một cá thể Kiến
Quá trình kiến xây dựng đường đi được lặp lại cho đến khi tất cả các môn học đều được xếp lịch, hoặc bị tắc nghẽn. Các bước cụ thể trong mỗi vòng lặp như sau:

#### 1. Chọn môn học và xử lý ràng buộc cứng
- Dựa trên heuristic ưu tiên, kiến chọn môn học $i$ tiếp theo để xếp lịch (hàm `_get_next_course` đã viết ở trên).
- Lọc ra tập các ca thi khả thi: Là những ca thi chưa bị chiếm bởi bất kỳ môn học nào có chung sinh viên với môn $i$ và đã được xếp lịch trước đó.

#### 2. Tính toán hàm heuristic cục bộ
- Với mỗi ca thi $j$ khả thi, tính toán độ phù hợp $\eta_{ij} = \frac{1}{1 + \Delta_\text{soft}}$.
- Chi phí gia tăng $\Delta_\text{soft}$ là tổng điểm phạt phát sinh nếu đặt môn học $i$ vào ca thi $j$. Điểm phạt đánh giá mức độ dày đặc của lịch thi, dựa trên khoảng cách ngày tuyệt đối với các môn học $k$ đã được xếp lịch trước đó có chung sinh viên: $$\Delta_\text{soft} = \sum_{k \in \text{Assigned}} c_{ik} \cdot 2^{-|d_j - d_{S_k}|}$$ trong đó:
  - $c_{ik}$: Số lượng sinh viên học cả hai môn $i$ và $k$.
  - $d_j$: Thời điểm của ca thi $j$ đang xem xét cho môn $i$, tính bằng ngày tuyệt đối, ví dụ 1.35 ngày.
  - $d_{S_k}$: Thời điểm của ca thi mà môn $k$ đã được xếp trước đó, tính bằng ngày tuyệt đối.

#### 3. Chọn ca thi
Xác suất kiến chọn ca thi $j$ cho môn $i$ tỷ lệ thuận với trọng số $w_{ij}$: $$w_{ij} \propto \tau_{ij}^\alpha \cdot \eta_{ij}^\beta$$ trong đó:
  - $\tau_{ij}$: Nồng độ pheromone hiện tại trên cạnh $(i, j)$
  - $\eta_{ij}$: Độ phù hợp cục bộ cục bộ nếu chọn ca thi $j$ cho môn $i$, đã tính ở bước trên.
  - $\alpha$: Tham số điều khiển mức độ ảnh hưởng của pheromone.
  - $\beta$: Tham số điều khiển mức độ ảnh hưởng của heuristic.

Có 2 hướng chọn ca thi:
- **Khai thác**: Luôn chọn ca thi có trọng số $w_{ij}$ lớn nhất.
- **Khám phá**: Kiến thực hiện quay Roulette để chọn ra đúng một ca thi $j$ dựa trên phân phối trọng số này.

#### 4. Cập nhật lịch thi và pheromone cục bộ
Sau khi môn $i$ được xếp vào ca $j$, nồng độ pheromone trên cạnh $(i, j)$ sẽ bị làm bay hơi một phần nhằm tránh làm cạnh này bị quá ưu tiên (tức là, khuyến khích các con kiến đi sau khám phá những cách xếp ca thi khác): $$\tau_{ij} \gets (1 - \xi) \cdot \tau_{ij} + \xi \cdot \tau_0$$ trong đó:
  - Hằng số $\xi$: Hệ số bay hơi cục bộ.
  - Hằng số $\tau_0$: Lượng pheromone khởi tạo.

```python
@njit
def construct_ant(
  tau: NDArray[np.float64],
  conflict: NDArray[np.int32] = CONFLICT,
  absolute_day_slots: NDArray[np.float64] = ABSOLUTE_DAY_SLOTS,
  num_courses: int = NUM_COURSES,
  num_slots: int = NUM_SLOTS,
  alpha: float = ALPHA,
  beta: float = BETA,
  xi: float = XI,
  tau_min: float = TAU_MIN,
  greedy_prob: float = GREEDY_PROB
) -> tuple[bool, NDArray[np.int32]]:

  # Khởi tạo các mảng 1 lần duy nhất ở ngoài vòng lặp
  # để tránh overhead cấp phát bộ nhớ
  schedule = np.full(num_courses, -1, dtype=np.int32)
  forbidden_mask = np.zeros((num_courses, num_slots), dtype=np.bool_)
  forbidden_count = np.zeros(num_courses, dtype=np.int32)
  feasible_slots = np.zeros(num_slots, dtype=np.int32)
  weights = np.zeros(num_slots, dtype=np.float64)
  conflicting_courses = np.zeros(num_courses, dtype=np.int32)
  conflicting_slots = np.zeros(num_courses, dtype=np.int32)

  for _ in range(num_courses):
    # Chọn môn học
    course = _get_next_course(schedule, forbidden_count)

    # Tìm các slot khả thi và kiểm tra có đang đi vào ngõ cụt không?
    num_feasible = 0
    for slot in range(num_slots):
      if not forbidden_mask[course, slot]:
        feasible_slots[num_feasible] = slot
        num_feasible += 1
    if num_feasible == 0:
      return False, schedule

    # Gom các môn xung đột đã xếp lịch
    num_conflicts = 0
    for course2 in range(num_courses):
      slot2 = schedule[course2]
      if slot2 != -1 and conflict[course, course2] > 0:
        conflicting_courses[num_conflicts] = course2
        conflicting_slots[num_conflicts] = slot2
        num_conflicts += 1

    # Duyệt từng slot khả thi và tính penalty nếu đặt môn học vào slot đó
    sum_w = 0.0
    best_j = 0
    for j in range(num_feasible):
      slot = feasible_slots[j]
      delta_soft = 0.0

      # Tính tổng delta_soft cho các môn xung đột đã xếp lịch
      for k in range(num_conflicts):
        course2 = conflicting_courses[k]
        slot2 = conflicting_slots[k]
        c_ik = conflict[course, course2]
        delta_days = absolute_day_slots[slot] - absolute_day_slots[slot2]
        delta_soft += c_ik * (2.0 ** -abs(delta_days))

      # Tính trọng số và tìm index của ca thi tốt nhất `best_j`
      eta = 1.0 / (1.0 + delta_soft)
      w = (tau[course, slot] ** alpha) * (eta ** beta)
      sum_w += w
      weights[j] = w
      if w > weights[best_j]:
        best_j = j

    # Chọn ca thi
    # Có 2 cách chọn ca thi là Khai thác và Khám phá như đã viết ở trên
    # Ở đây thực hiện cách Khám phá, còn việc tìm ra `best_j` ở trên
    # đã là phục vụ cho cách Khai thác
    if np.random.random() >= greedy_prob:
      threshold = np.random.random() * sum_w
      cumulative = 0.0
      for j in range(num_feasible):
        cumulative += weights[j]
        if cumulative >= threshold:
          best_j = j
          break

    # Cập nhật lịch thi và pheromone cục bộ
    slot = feasible_slots[best_j]
    schedule[course] = slot
    tau[course, slot] = (1 - xi) * tau[course, slot] + xi * tau_min

    _update_forbidden_mask(course, slot, schedule, forbidden_mask, forbidden_count)
  
  return True, schedule
```

### Tính độ thay đổi cost khi chuyển lịch thi của một môn sang slot khác

```python
@njit
def _evaluate_delta(
  schedule: NDArray[np.int32],
  course: int,
  new_slot: int,
  ignore_course: int = -1,
  num_courses: int = NUM_COURSES,
  conflict: NDArray[np.int32] = CONFLICT,
  absolute_day_slots: NDArray[np.float64] = ABSOLUTE_DAY_SLOTS
) -> float:

  d_old = absolute_day_slots[schedule[course]]
  d_new = absolute_day_slots[new_slot]
  delta = 0.0

  for course2 in range(num_courses):
    weight = conflict[course, course2]
    if weight == 0 or course2 == ignore_course or course2 == course:
      continue
    d_S_k = absolute_day_slots[schedule[course2]]
    penalty_old = weight * (2.0 ** -abs(d_old - d_S_k))
    penalty_new = weight * (2.0 ** -abs(d_new - d_S_k))
    delta += penalty_new - penalty_old

  return delta
```

### Tính tổng cost của toàn bộ lịch thi

```python
@njit
def evaluate_cost(
  schedule: NDArray[np.int32],
  num_courses: int = NUM_COURSES,
  conflict: NDArray[np.int32] = CONFLICT,
  absolute_day_slots: NDArray[np.float64] = ABSOLUTE_DAY_SLOTS
) -> float:
  
  total_cost = 0.0
  for course1 in range(num_courses):
    d_i = absolute_day_slots[schedule[course1]]
    for course2 in range(course1 + 1, num_courses):
      weight = conflict[course1, course2]
      if weight == 0:
        continue
      d_j = absolute_day_slots[schedule[course2]]
      total_cost += weight * (2.0 ** -abs(d_i - d_j))

  if total_cost < 0:
    print("negative cost")
  return total_cost
```

### Local Search: 1-move

```python
@njit
def _go_1_move(
  schedule: NDArray[np.int32],
  forbidden_mask_all: NDArray[np.bool_],
  feasible_slots: NDArray[np.int32],
  num_courses: int = NUM_COURSES,
  num_slots: int = NUM_SLOTS,
  conflict: NDArray[np.int32] = CONFLICT
) -> tuple[bool, float]:

  # Chọn ngẫu nhiên một môn học
  course = np.random.randint(0, num_courses)

  # Reset forbidden_mask của môn đang xét
  forbidden_mask = forbidden_mask_all[course]
  forbidden_mask.fill(False)
  forbidden_mask[schedule[course]] = True

  # Tìm các slot bị cấm
  for course2 in range(num_courses):
    slot2 = schedule[course2]
    if conflict[course, course2] > 0:
      forbidden_mask[slot2] = True

  # Tìm các slot khả thi
  num_feasible = 0
  for slot in range(num_slots):
    if not forbidden_mask[slot]:
      feasible_slots[num_feasible] = slot
      num_feasible += 1
  if num_feasible == 0:
    return False, float('inf')

  # Chọn ngẫu nhiên một slot, tính độ thay đổi cost khi chuyển lịch thi sang slot mới
  # và chấp nhận thay đổi nếu cost giảm
  new_slot = feasible_slots[np.random.randint(0, num_feasible)]
  delta = _evaluate_delta(schedule, course, new_slot)
  if delta <= 0:
    schedule[course] = new_slot
    return True, delta
  return False, delta
```

### Local Search: 2-swap

```python
@njit
def _go_2_swap(
  schedule: NDArray[np.int32],
  num_courses: int = NUM_COURSES,
  conflict: NDArray[np.int32] = CONFLICT
) -> tuple[bool, float]:

  # Chọn ngẫu nhiên hai môn học
  course1 = np.random.randint(0, num_courses)
  course2 = np.random.randint(0, num_courses)
  slot1 = schedule[course1]
  slot2 = schedule[course2]

  # Kiểm tra ràng buộc cứng khi hoán đổi slot với nhau
  if course1 == course2 or slot1 == slot2:
    return False, float('inf')
  for c in range(num_courses):
    if c == course1 or c == course2:
      continue
    if (schedule[c] == slot2 and conflict[course1, c] > 0) or \
       (schedule[c] == slot1 and conflict[course2, c] > 0):
      return False, float('inf')

  # Tính độ thay đổi cost khi hoán đổi slot với nhau
  # và chấp nhận thay đổi nếu cost giảm
  delta = (_evaluate_delta(schedule, course1, slot2, course2)
         + _evaluate_delta(schedule, course2, slot1, course1))
  if delta <= 0:
    schedule[course1] = slot2
    schedule[course2] = slot1
    return True, delta
  return False, delta
```

### Local Search: main function

```python
@njit
def local_search(
  schedule: NDArray[np.int32],
  num_courses: int = NUM_COURSES,
  num_slots: int = NUM_SLOTS,
  max_moves: int = MAX_MOVES,
  max_fails: int = MAX_FAILS,
  move_prob: float = MOVE_PROB
) -> float:

  forbidden_mask = np.zeros((num_courses, num_slots), dtype=np.bool_)
  feasible_slots = np.zeros(num_slots, dtype=np.int32)
  delta_cost = 0.0
  moves_count = 0
  consecutive_fails_count = 0

  while moves_count < max_moves and consecutive_fails_count < max_fails:
    if np.random.random() < move_prob:
      is_moved, delta = _go_1_move(schedule, forbidden_mask, feasible_slots)
    else:
      is_moved, delta = _go_2_swap(schedule)

    if is_moved:
      delta_cost += delta
      moves_count += 1
      consecutive_fails_count = 0
    else:
      consecutive_fails_count += 1

  return delta_cost
```

### Cập nhật pheromone toàn cục
Cập nhật pheromone cho mọi cặp (course, slot) $(i, j)$ dựa trên thuật toán Smoothed Max-Min Ant System (SMMAS). Quá trình này giúp điều chỉnh nồng độ pheromone tiệm cận một cách tự nhiên trong khoảng giới hạn $[\tau_{\min}, \tau_{\max}]$ để tránh một vài đường đi bị quá ưu tiên hoặc bị hội tụ sớm: $$\tau_{ij} = (1 - \rho)\tau_{ij} + \Delta\tau_{ij}$$ Trong đó, lượng pheromone cộng thêm $\Delta\tau_{ij}$ được xác định dựa trên việc cặp $(i, j)$ đó có thuộc về lời giải tốt nhất `best_schedule` hay không:
- Deposit: Nếu cặp $(i, j)$ thuộc `best_schedule`: $\Delta\tau_{ij} = \rho\cdot\tau_{\max}$
- Evaporate: Nếu cặp $(i, j)$ không thuộc `best_schedule`: $\Delta\tau_{ij} = \rho\cdot\tau_{\min}$

```python
@njit
def evaporate_and_deposit(
  tau: NDArray[np.float64],
  best_schedule: NDArray[np.int32],
  num_courses: int = NUM_COURSES,
  num_slots: int = NUM_SLOTS,
  rho: float = RHO,
  tau_min: float = TAU_MIN,
  tau_max: float = TAU_MAX
) -> None:

  for course in range(num_courses):
    for slot in range(num_slots):
      if slot == best_schedule[course]:
        delta_tau = rho * tau_max
      else:
        delta_tau = rho * tau_min
      tau[course, slot] = (1.0 - rho) * tau[course, slot] + delta_tau
```

### Main algorithm
Ant colony optimization + Local search

```python
best_schedule = np.full(NUM_COURSES, -1, dtype=np.int32)
global_best = np.float64(np.inf)
no_improvement = 0

tau = np.full((NUM_COURSES, NUM_SLOTS), TAU_0, dtype=np.float64)
history_global_best = []
history_iter_best = []
fig = plt.figure(figsize=(10, 5))

for iter in range(MAX_ITERS):
  ants = np.zeros((NUM_ANTS, NUM_COURSES), dtype=np.int32)
  costs = np.zeros(NUM_ANTS, dtype=np.float64)
  j = 0
  while j < NUM_ANTS:
    ok, schedule = construct_ant(tau)
    if not ok:
      continue
    ants[j] = schedule
    costs[j] = evaluate_cost(schedule)
    j += 1
  sorted_indices = np.argsort(costs)
  ants = ants[sorted_indices]
  costs = costs[sorted_indices]

  num_elites = round(NUM_ANTS * TOP_ANTS)
  for j in range(num_elites):
    costs[j] += local_search(ants[j])

  i_best = np.argmin(costs[:num_elites])
  if costs[i_best] < global_best:
    best_schedule = ants[i_best]
    global_best = costs[i_best]
    no_improvement = 0
  else:
    no_improvement += 1
    if no_improvement == STAGNATION_LIMIT:
      # tau[:] = TAU_0 + 0.3 * (tau - TAU_0)
      # np.clip(tau, TAU_MIN, TAU_MAX, out=tau)
      tau.fill(TAU_0)
      no_improvement = 0
  # if no_improvement < 10:
  # evaporate_and_deposit(tau, best_schedule)
  # else:
  evaporate_and_deposit(tau, ants[i_best])

  history_global_best.append(global_best)
  history_iter_best.append(costs[i_best])
  clear_output(wait=True)
  plt.figure(figsize=(12, 10))
        
  # Đường màu cam: Điểm tốt nhất của từng vòng lặp (Nhảy múa liên tục thể hiện sự khám phá)
  plt.plot(history_iter_best, label='Iteration Best', color='orange', linewidth=0.5)
  
  # Đường màu xanh: Điểm tốt nhất toàn cục (Đường tiệm cận đi xuống)
  plt.plot(history_global_best, label='Global Best', color='blue', linewidth=2)
  
  plt.title(f"ACO Optimization Progress - Iteration {iter + 1} | Best Cost: {global_best:.2f}")
  plt.xlabel("Iteration")
  plt.ylabel("Cost (Penalty)")
  plt.legend()
  plt.grid(True, linestyle=':', alpha=0.7)
  
  plt.show()
```

### Gán phòng thi và cán bộ coi thi
Sau khi đã có lịch thi tối ưu nhất, ta sẽ gán phòng thi và cán bộ coi thi.  
Đối với mỗi ca thi $i$, gọi $n$ là số lượng sinh viên thi ca này. Giả sử mỗi phòng có tối đa $40$ sinh viên và mỗi phòng thi cần đúng $2$ cán bộ coi thi, ta cần $r = \lceil \frac{n}{40} \rceil$ phòng thi và $t = 2r$ cán bộ coi thi.  
Về vấn đề phòng thi, ta có thể xếp môn thi vào bất cứ phòng nào miễn là phòng đó không được sử dụng bởi môn thi khác trong cùng ca thi.  
Về vấn đề cán bộ coi thi, ta mong muốn các cán bộ coi thi được phân việc đồng đều nhất (không có cán bộ nào phải đi coi quá nhiều ca thi trong khi có cán bộ coi ít ca thi hơn).

```python
N_SUPERVISORS = 140
N_ROOMS       = 70
ROOM_MAX_CAPACITY = 40

schedule_rows: list[tuple[str, str, str, int, int, int, int]] = []
count_timeslots_of_supervisor = [0] * N_SUPERVISORS
schedule_by_timeslot: dict[int, list[str]] = defaultdict(list)
for course, ts in best_schedule.items():
  schedule_by_timeslot[ts].append(course)
for ts, COURSES in schedule_by_timeslot.items():
  supervisors = sorted(range(N_SUPERVISORS), key=lambda sp: count_timeslots_of_supervisor[sp])
  day, timeslot = TIMESLOTS[ts]
  room_id = 0
  for course in COURSES:
    # Sort theo mã lớp môn học vì ta muốn có nhiều sinh viên học cùng lớp thì thi cùng phòng nhất chứ không phải xếp ngẫu nhiên
    students = sorted(students_by_course[course], key=lambda sid: get_section_of_course(sid, course))
    n_rooms = ceil(len(students) / ROOM_MAX_CAPACITY)
    idx = 0
    for j in range(n_rooms):
      # Tránh trường hợp có phòng maximum sinh viên trong khi có phòng có đúng 1 sinh viên, ta mong muốn mỗi phòng có số sinh viên bằng nhau
      # Ví dụ có 10 sv và 3 phòng thi thì có 1 phòng 4 sv và 2 phòng 3 sv
      capacity = len(students) // n_rooms + (j < len(students) % n_rooms)
      sp1 = supervisors[room_id * 2]
      sp2 = supervisors[room_id * 2 + 1]
      count_timeslots_of_supervisor[sp1] += 1
      count_timeslots_of_supervisor[sp2] += 1
      for j in range(capacity):
        daystr = (DATE_START + timedelta(days=day)).strftime(r"%Y-%m-%d")
        schedule_rows.append((students[idx], course, daystr, timeslot % SLOTS_PER_DAY + 1, room_id + 1, sp1 + 1, sp2 + 1))
        idx += 1
      room_id += 1
```
